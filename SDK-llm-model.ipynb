```python
# with function
from google import genai
from google.genai import types

client = genai.Client(api_key = "api_key",
                       http_options=types.HttpOptions(
        timeout=30000,
        retry_options=types.HttpRetryOptions(attempts=1)
    ))
model = "gemini-3.6-flash"

def ask_gemini(prompt):
    interaction = client.interactions.create(
        model = model,
        input = prompt)
    return interaction.output_text
try:
    result = ask_gemini("what is your cutoff date")
    print(result)
except Exception as e:
    print(e)
```

```python
result1 = ask_gemini("what is python in one line?")
result2 = ask_gemini("what is java in one line?")
result3 = ask_gemini("what is aws in one line?")
print(result1)
print(result2)
print(result3)
```

```python
from google import genai
from google.genai import types
import time

print("1. Starting...", flush=True)

client = genai.Client(
    api_key="api_key",
    http_options=types.HttpOptions(
        timeout=30000,
        retry_options=types.HttpRetryOptions(
            attempts=1
        )
    )
)

print("2. Client created", flush=True)

start = time.perf_counter()

try:
    response = client.interactions.create(
        model="gemini-3.6-flash",
        input="Say hello"
    )

    print("Response:", response.output_text)

except Exception as e:
    print("API ERROR:", type(e).__name__)
    print("Details:", str(e))

finally:
    print("Execution time:",
          round(time.perf_counter() - start, 2), "seconds")
```
